```python
import pandas as pd
import numpy as np
import os
import pickle
from collections import Counter
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import precision_score, accuracy_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import xgboost as xgb
import lightgbm as lgb
from imblearn.combine import SMOTEENN  # ← CHANGE 1: SMOTE+ENN
import matplotlib.pyplot as plt
import time

# Setup paths
data_dir = r'C:\Users\LIE KHUAN HUNG\OneDrive\Desktop\fyp\elliptic_bitcoin_dataset'
np.random.seed(42)

# Load data
features_df = pd.read_csv(os.path.join(data_dir, 'elliptic_txs_features.csv'), header=None)
features_df.columns = ['txId', 'timestamp'] + [f'feature_{i}' for i in range(1, 166)]
classes_df = pd.read_csv(os.path.join(data_dir, 'elliptic_txs_classes.csv'))
df = pd.merge(features_df, classes_df, on='txId')
df = df[df['class'] != 'unknown']
df['class'] = df['class'].map({'1': 1, '2': 0})
df = df.dropna(subset=['class'])

# Feature matrix, labels, and timestamps
X_all = df.drop(columns=['txId', 'class']).values
y_all = df['class'].values
timestamps_all = df['timestamp'].values

# Normalize
scaler = MinMaxScaler()
X_all_scaled = scaler.fit_transform(X_all)
print("Filtered shape:", X_all_scaled.shape)
print("Class distribution:\n", pd.Series(y_all).value_counts(normalize=True))

# ▶️ Ensemble Feature Selection
rf_model = RandomForestClassifier(random_state=42)
xgb_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
lgb_model = lgb.LGBMClassifier(random_state=42)
print("Training RF/XGB/LGB for feature selection...")
rf_model.fit(X_all_scaled, y_all)
xgb_model.fit(X_all_scaled, y_all)
lgb_model.fit(X_all_scaled, y_all)

# Remove timestamp (index 0), get importances
rf_importance = rf_model.feature_importances_[1:]
xgb_importance = xgb_model.feature_importances_[1:]
lgb_importance = lgb_model.feature_importances_[1:]

top_rf = np.argsort(rf_importance)[::-1][:50]
top_xgb = np.argsort(xgb_importance)[::-1][:50]
top_lgb = np.argsort(lgb_importance)[::-1][:50]

all_indices = np.concatenate([top_rf, top_xgb, top_lgb])
vote_counts = Counter(all_indices)

# Select top 20, 30, 40 features
feature_ranges = [20, 30, 40]
feature_indices = {}
for n in feature_ranges:
    common_features = [idx for idx, _ in vote_counts.most_common(n)]
    feature_indices[n] = np.array([i + 1 for i in common_features])  # +1 to account for timestamp exclusion
    print(f"Top {n} selected feature indices (excluding timestamp):", feature_indices[n])

# ▶️ Centralize Data Splitting and SMOTE+ENN
splits = {'all': {}, 'top_20': {}, 'top_30': {}, 'top_40': {}}
for feature_set in splits:
    if feature_set == 'all':
        X = X_all_scaled
    else:
        n = int(feature_set.split('_')[1])
        X = X_all_scaled[:, feature_indices[n]]
    # Split with stratification, including timestamps
    X_train, X_test, y_train, y_test, ts_train, ts_test = train_test_split(
        X, y_all, timestamps_all, test_size=0.2, stratify=y_all, random_state=42
    )
    # Apply SMOTE+ENN ← CHANGE 1
    smote_enn = SMOTEENN(sampling_strategy=0.3, random_state=42)
    X_train_res, y_train_res = smote_enn.fit_resample(X_train, y_train)
    # Store splits
    splits[feature_set] = {
        'X_train': X_train, 'X_test': X_test,
        'y_train': y_train, 'y_test': y_test,
        'X_train_res': X_train_res, 'y_train_res': y_train_res,
        'ts_test': ts_test
    }

# ▶️ Define the pipeline
def run_pipeline(X_train_res, y_train_res, X_test, y_test, feature_set='all'):
    # Class weight for XGB, LGBM
    class_counts = pd.Series(y_train_res).value_counts()
    scale_pos_weight = (class_counts[0] / class_counts[1]) * 1.5
    models = {
        'RandomForest': RandomForestClassifier(class_weight='balanced', random_state=42),
        'XGBoost': xgb.XGBClassifier(scale_pos_weight=scale_pos_weight, use_label_encoder=False, eval_metric='logloss', random_state=42),
        'LightGBM': lgb.LGBMClassifier(scale_pos_weight=scale_pos_weight, random_state=42)
    }
    param_grids = {
        'RandomForest': {
            'max_depth': [5, 8],
            'min_samples_split': [5, 10],
            'min_samples_leaf': [1, 5],
            'n_estimators': [100, 200]
        },
        'XGBoost': {
            'max_depth': [5, 7],
            'min_child_weight': [3, 5],
            'n_estimators': [100, 200],
            'learning_rate': [0.05, 0.1],
        },
        'LightGBM': {
            'max_depth': [5, 7],
            'learning_rate': [0.05, 0.1],
            'n_estimators': [100, 200]
        }
    }
    thresholds = [0.05, 0.1, 0.2, 0.3]
    results = {'Model': [], 'Train_Precision': [], 'Train_Accuracy': [], 'Test_Precision': [], 'Test_Accuracy': [], 'Test_Time': [], 'Best_Threshold': []}
    for model_name in models:
        print(f"\nTuning {model_name} for {feature_set}...")
        checkpoint_file = os.path.join(data_dir, f"{model_name.lower()}_checkpoint_{feature_set}.pkl")
       
        # Check if checkpoint exists
        if os.path.exists(checkpoint_file):
            print(f"Loading checkpoint for {model_name} from {checkpoint_file}")
            with open(checkpoint_file, 'rb') as f:
                checkpoint = pickle.load(f)
                model = checkpoint['model']
                best_t = checkpoint['best_threshold']
                best_precision = checkpoint['best_precision']
                best_accuracy = checkpoint['best_accuracy']
            print(f"Loaded best threshold: {best_t}, Precision: {best_precision:.4f}, Accuracy: {best_accuracy:.4f}")
        else:
            # ← CHANGE 2: Use average_precision
            grid = GridSearchCV(
                models[model_name], param_grids[model_name],
                scoring='average_precision',  # ← BETTER FOR IMBALANCED DATA
                cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
                n_jobs=-1, verbose=1
            )
            grid.fit(X_train_res, y_train_res)
            model = grid.best_estimator_
            best_params = grid.best_params_  # ← ADD THIS LINE**
            best_t, best_precision, best_accuracy = 0.5, 0, 0
            for t in thresholds:
                precisions, accuracies = [], []
                for train_idx, val_idx in StratifiedKFold(n_splits=5, shuffle=True, random_state=42).split(X_train_res, y_train_res):
                    model.fit(X_train_res[train_idx], y_train_res[train_idx])
                    preds = (model.predict_proba(X_train_res[val_idx])[:, 1] > t).astype(int)
                    precisions.append(precision_score(y_train_res[val_idx], preds, zero_division=0))
                    accuracies.append(accuracy_score(y_train_res[val_idx], preds))
                avg_precision = np.mean(precisions)
                avg_accuracy = np.mean(accuracies)
                if avg_precision > 0.95 and avg_accuracy > best_accuracy:
                    best_t, best_precision, best_accuracy = t, avg_precision, avg_accuracy
            # Save checkpoint
            checkpoint = {'model': model, 'best_threshold': best_t, 'best_precision': best_precision, 'best_accuracy': best_accuracy, 'best_params': best_params}
            with open(checkpoint_file, 'wb') as f:
                pickle.dump(checkpoint, f)
            print(f"Saved checkpoint for {model_name} to {checkpoint_file}")
            print(f"Best threshold: {best_t}, Precision: {best_precision:.4f}, Accuracy: {best_accuracy:.4f}")
       
        # Train with best parameters
        model.fit(X_train_res, y_train_res)
       
        # Predict and time it
        start_time = time.time()
        train_preds = (model.predict_proba(X_train_res)[:, 1] > best_t).astype(int)
        test_preds = (model.predict_proba(X_test)[:, 1] > best_t).astype(int)
        test_time = time.time() - start_time
        results['Model'].append(model_name)
        results['Train_Precision'].append(precision_score(y_train_res, train_preds, zero_division=0))
        results['Train_Accuracy'].append(accuracy_score(y_train_res, train_preds))
        results['Test_Precision'].append(precision_score(y_test, test_preds, zero_division=0))
        results['Test_Accuracy'].append(accuracy_score(y_test, test_preds))
        results['Test_Time'].append(test_time)
        results['Best_Threshold'].append(best_t)
    return pd.DataFrame(results)

# ▶️ Run experiments for all features and feature ranges
results = {}
for feature_set in splits:
    print(f"\nRunning Experiment - {feature_set}")
    results[feature_set] = run_pipeline(
        splits[feature_set]['X_train_res'], splits[feature_set]['y_train_res'],
        splits[feature_set]['X_test'], splits[feature_set]['y_test'], feature_set
    )

# ▶️ Combined Test Performance Table (Replaced separate tables)
combined_data = []
fs_map = {
    'all': 'All (166)',
    'top_20': 'Top 20',
    'top_30': 'Top 30',
    'top_40': 'Top 40'
}
for fs_key, res_df in results.items():
    fs_name = fs_map[fs_key]
    temp_df = res_df[['Model', 'Test_Precision', 'Test_Accuracy', 'Test_Time']].copy()
    temp_df.columns = ['Model', 'Precision', 'Accuracy', 'Test Time (s)']
    temp_df['Features'] = fs_name
    combined_data.append(temp_df[['Features', 'Model', 'Precision', 'Accuracy', 'Test Time (s)']])

combined_df = pd.concat(combined_data, ignore_index=True)
combined_df = combined_df.round(4)
combined_df = combined_df.set_index(['Features', 'Model'])

print("\nTest Performance - All Feature Subsets")
display(combined_df.style
        .set_caption("Model Performance on Test Set - All Feature Subsets")
        .format(precision=4)
        .background_gradient(cmap='Blues', axis=0))

# ▶️ Precision Plot
models = results['all']['Model']
x = np.arange(len(models))
width = 0.2
plt.figure(figsize=(12, 6))
plt.bar(x - 1.5*width, results['all']['Test_Precision'], width, label='All Features', color='skyblue')
plt.bar(x - 0.5*width, results['top_20']['Test_Precision'], width, label='Top 20 Features', color='green')
plt.bar(x + 0.5*width, results['top_30']['Test_Precision'], width, label='Top 30 Features', color='purple')
plt.bar(x + 1.5*width, results['top_40']['Test_Precision'], width, label='Top 40 Features', color='red')
for i, v in enumerate(results['all']['Test_Precision']):
    plt.text(i - 1.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
for i, v in enumerate(results['top_20']['Test_Precision']):
    plt.text(i - 0.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
for i, v in enumerate(results['top_30']['Test_Precision']):
    plt.text(i + 0.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
for i, v in enumerate(results['top_40']['Test_Precision']):
    plt.text(i + 1.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
plt.ylim(min(min(results['all']['Test_Precision']), min(results['top_20']['Test_Precision']),
             min(results['top_30']['Test_Precision']), min(results['top_40']['Test_Precision'])) - 0.01, 1.01)
plt.xticks(x, models)
plt.ylabel('Precision')
plt.title('Test Precision Comparison')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# ▶️ Accuracy Plot
plt.figure(figsize=(12, 6))
plt.bar(x - 1.5*width, results['all']['Test_Accuracy'], width, label='All Features', color='skyblue')
plt.bar(x - 0.5*width, results['top_20']['Test_Accuracy'], width, label='Top 20 Features', color='green')
plt.bar(x + 0.5*width, results['top_30']['Test_Accuracy'], width, label='Top 30 Features', color='purple')
plt.bar(x + 1.5*width, results['top_40']['Test_Accuracy'], width, label='Top 40 Features', color='red')
for i, v in enumerate(results['all']['Test_Accuracy']):
    plt.text(i - 1.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
for i, v in enumerate(results['top_20']['Test_Accuracy']):
    plt.text(i - 0.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
for i, v in enumerate(results['top_30']['Test_Accuracy']):
    plt.text(i + 0.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
for i, v in enumerate(results['top_40']['Test_Accuracy']):
    plt.text(i + 1.5*width, v + 0.001, f'{v:.4f}', ha='center', va='bottom')
plt.ylim(min(min(results['all']['Test_Accuracy']), min(results['top_20']['Test_Accuracy']),
             min(results['top_30']['Test_Accuracy']), min(results['top_40']['Test_Accuracy'])) - 0.01, 1.01)
plt.xticks(x, models)
plt.ylabel('Accuracy')
plt.title('Test Accuracy Comparison')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# ▶️ Testing Time Plot
plt.figure(figsize=(12, 6))
plt.bar(x - 1.5*width, results['all']['Test_Time'], width, label='All Features', color='skyblue')
plt.bar(x - 0.5*width, results['top_20']['Test_Time'], width, label='Top 20 Features', color='green')
plt.bar(x + 0.5*width, results['top_30']['Test_Time'], width, label='Top 30 Features', color='purple')
plt.bar(x + 1.5*width, results['top_40']['Test_Time'], width, label='Top 40 Features', color='red')
plt.xticks(x, models)
plt.ylabel('Testing Time (seconds)')
plt.title('Testing Time Comparison')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# ▶️ NEW: Option A Line Plots (Feature Subset on X-axis, Separate Lines for Models)
feature_subsets = ['All', 'Top 20', 'Top 30', 'Top 40']
subset_keys = ['all', 'top_20', 'top_30', 'top_40']

# Prepare data for line plots
metrics = {
    'Precision': [],
    'Accuracy': [],
    'Test Time': []
}
for model in ['RandomForest', 'XGBoost', 'LightGBM']:
    prec = [results[key][results[key]['Model'] == model]['Test_Precision'].values[0] for key in subset_keys]
    acc = [results[key][results[key]['Model'] == model]['Test_Accuracy'].values[0] for key in subset_keys]
    time_vals = [results[key][results[key]['Model'] == model]['Test_Time'].values[0] for key in subset_keys]
    metrics['Precision'].append(prec)
    metrics['Accuracy'].append(acc)
    metrics['Test Time'].append(time_vals)

# Function to create line plot
def create_line_plot(metric_name, y_label):
    plt.figure(figsize=(10, 6))
    for i, model in enumerate(['RandomForest', 'XGBoost', 'LightGBM']):
        plt.plot(feature_subsets, metrics[metric_name][i], marker='o', label=model)
    plt.xlabel('Feature Subset')
    plt.ylabel(y_label)
    plt.title(f'{metric_name} Across Feature Subsets')
    if metric_name in ['Precision', 'Accuracy']:
        all_values = [val for sublist in metrics[metric_name] for val in sublist]
        min_val = min(all_values)
        max_val = max(all_values)
        margin = (max_val - min_val) * 0.05 if (max_val - min_val) > 0 else 0.01
        plt.ylim(min_val - margin, max_val + margin)
    else:  # Test Time
        all_values = [val for sublist in metrics[metric_name] for val in sublist]
        min_val = min(all_values)
        max_val = max(all_values)
        margin = (max_val - min_val) * 0.05 if (max_val - min_val) > 0 else 0.01
        plt.ylim(min_val - margin, max_val + margin)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()

# Generate the three line plots
create_line_plot('Precision', 'Precision')
create_line_plot('Accuracy', 'Accuracy')
create_line_plot('Test Time', 'Test Time (seconds)')

# ▶️ Helper: Get model with best parameters
def get_model(name, spw):
    if name == 'RandomForest':
        return RandomForestClassifier(class_weight='balanced', max_depth=8, min_samples_split=5,
                                      min_samples_leaf=1, n_estimators=200, random_state=42)
    elif name == 'XGBoost':
        return xgb.XGBClassifier(scale_pos_weight=spw, max_depth=5, min_child_weight=3,
                                 n_estimators=200, learning_rate=0.05, use_label_encoder=False,
                                 eval_metric='logloss', random_state=42)
    elif name == 'LightGBM':
        return lgb.LGBMClassifier(scale_pos_weight=spw, max_depth=5, learning_rate=0.05,
                                  n_estimators=200, random_state=42)

# ▶️ Confusion Matrices
models_to_run = ['RandomForest', 'XGBoost', 'LightGBM']
for feature_set in splits:
    if feature_set == 'all':
        caption = "All Features"
    else:
        n = int(feature_set.split('_')[1])
        caption = f"Top {n} Features"
    print(f"\nConfusion Matrices - {caption}")
    plt.figure(figsize=(15, 4))
    for i, model_name in enumerate(models_to_run):
        spw = (pd.Series(splits[feature_set]['y_train_res']).value_counts()[0] /
               pd.Series(splits[feature_set]['y_train_res']).value_counts()[1]) * 1.5
        model = get_model(model_name, spw)
        model.fit(splits[feature_set]['X_train_res'], splits[feature_set]['y_train_res'])
        # Use best threshold from run_pipeline
        best_t = results[feature_set][results[feature_set]['Model'] == model_name]['Best_Threshold'].iloc[0]
        y_pred = (model.predict_proba(splits[feature_set]['X_test'])[:, 1] > best_t).astype(int)
        cm = confusion_matrix(splits[feature_set]['y_test'], y_pred)
        disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Licit (0)', 'Illicit (1)'])
        plt.subplot(1, 3, i + 1)
        disp.plot(ax=plt.gca(), cmap='Blues', values_format='d')
        plt.title(f'{model_name} (Threshold: {best_t:.2f})')
    plt.tight_layout()
    plt.show()

# ▶️ Updated Temporal Precision Plot
def temporal_precision_plot(X_train_res, y_train_res, X_test, y_test, ts_test, title_label, spw, thresholds, use_fixed_threshold=False, min_samples=5):
    temporal_results = {model: {} for model in models_to_run}
    fixed_threshold = 0.2
    for model_name in models_to_run:
        model = get_model(model_name, spw)
        model.fit(X_train_res, y_train_res)
        best_t = fixed_threshold if use_fixed_threshold else thresholds[model_name]
        y_pred = (model.predict_proba(X_test)[:, 1] > best_t).astype(int)
        for ts in np.unique(ts_test):
            ts_mask = ts_test == ts
            if np.sum(ts_mask) >= min_samples:
                y_true_ts = y_test[ts_mask]
                y_pred_ts = y_pred[ts_mask]
                if len(np.unique(y_true_ts)) > 1 and np.sum(y_true_ts == 1) >= 1:
                    f1 = f1_score(y_true_ts, y_pred_ts, zero_division=0)
                    print(f"Timestamp {ts} ({title_label}): {np.sum(y_true_ts == 1)} illicit samples, F1-score = {f1:.4f}")
                    temporal_results[model_name][ts] = f1
                else:
                    print(f"Timestamp {ts} ({title_label}): Skipped (no illicit samples or single class)")
    plt.figure(figsize=(12, 6))
    for model_name in models_to_run:
        timestamps_sorted = sorted(temporal_results[model_name].keys())
        if timestamps_sorted:
            f1_scores = [temporal_results[model_name][ts] for ts in timestamps_sorted]
            plt.plot(timestamps_sorted, f1_scores, marker='o', label=model_name)
        else:
            print(f"No valid F1-scores for {model_name} in {title_label}")
    plt.axhline(y=0, color='gray', linestyle='-', linewidth=0.5)
    plt.ylim(-0.05, 1.05)
    plt.xlabel("Timestamp")
    plt.ylabel("Illicit F1-Score")
    plt.title(f"Illicit F1-Score Over Timestamps - {title_label}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

# Run temporal precision plots
for feature_set in splits:
    if feature_set == 'all':
        title_label = "All Features"
    else:
        n = int(feature_set.split('_')[1])
        title_label = f"Top {n} Features"
    spw = (pd.Series(splits[feature_set]['y_train_res']).value_counts()[0] /
           pd.Series(splits[feature_set]['y_train_res']).value_counts()[1]) * 1.5
    thresholds = {row['Model']: row['Best_Threshold'] for _, row in results[feature_set].iterrows()}
    temporal_precision_plot(
        splits[feature_set]['X_train_res'], splits[feature_set]['y_train_res'],
        splits[feature_set]['X_test'], splits[feature_set]['y_test'],
        splits[feature_set]['ts_test'], title_label, spw, thresholds,
        use_fixed_threshold=True, min_samples=5
    )

# ▶️ Illicit Distribution Plot
def plot_illicit_distribution_multi(X_train_res, y_train_res, X_test, y_test, ts_test, title_label, spw, thresholds, use_fixed_threshold=False):
    timestamps_sorted = sorted(np.unique(ts_test))
    fixed_threshold = 0.2
    true_illicit = pd.Series(y_test).groupby(ts_test).sum()
    true_counts = [true_illicit.get(ts, 0) for ts in timestamps_sorted]
    plt.figure(figsize=(12, 6))
    plt.plot(timestamps_sorted, true_counts, label='True Illicit', marker='o', color='black')
    for model_name in models_to_run:
        model = get_model(model_name, spw)
        model.fit(X_train_res, y_train_res)
        best_t = fixed_threshold if use_fixed_threshold else thresholds[model_name]
        y_pred = (model.predict_proba(X_test)[:, 1] > best_t).astype(int)
        pred_illicit = pd.Series(y_pred).groupby(ts_test).sum()
        pred_counts = [pred_illicit.get(ts, 0) for ts in timestamps_sorted]
        plt.plot(timestamps_sorted, pred_counts, linestyle='--', marker='x', label=f'{model_name} Predicted')
    plt.xlabel("Timestamp")
    plt.ylabel("Number of Illicit Transactions")
    plt.title(f"Illicit Node Distribution Over Timestamps - {title_label}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

# Run illicit distribution plots
for feature_set in splits:
    if feature_set == 'all':
        title_label = "All Features"
    else:
        n = int(feature_set.split('_')[1])
        title_label = f"Top {n} Features"
    spw = (pd.Series(splits[feature_set]['y_train_res']).value_counts()[0] /
           pd.Series(splits[feature_set]['y_train_res']).value_counts()[1]) * 1.5
    thresholds = {row['Model']: row['Best_Threshold'] for _, row in results[feature_set].iterrows()}
    plot_illicit_distribution_multi(
        splits[feature_set]['X_train_res'], splits[feature_set]['y_train_res'],
        splits[feature_set]['X_test'], splits[feature_set]['y_test'],
        splits[feature_set]['ts_test'], title_label, spw, thresholds,
        use_fixed_threshold=True
    )

    # ▶️ Best Hyperparameters Table (NEW!)
fs_display = {
    'all': 'All Features', 'top_20': 'Top 20 Features',
    'top_30': 'Top 30 Features', 'top_40': 'Top 40 Features'
}
model_display = {'randomforest': 'RandomForest', 'xgboost': 'XGBoost', 'lightgbm': 'LightGBM'}
param_keys = {
    'randomforest': ['max_depth', 'min_samples_split', 'min_samples_leaf', 'n_estimators'],
    'xgboost': ['max_depth', 'min_child_weight', 'n_estimators', 'learning_rate'],
    'lightgbm': ['max_depth', 'learning_rate', 'n_estimators']
}

best_hyper_table = []
for fs in fs_display:
    for m_lower, m_disp in model_display.items():
        ckpt_file = os.path.join(data_dir, f"{m_lower}_checkpoint_{fs}.pkl")
        if os.path.exists(ckpt_file):
            with open(ckpt_file, 'rb') as f:
                ckpt = pickle.load(f)
            best_t = ckpt['best_threshold']
            
            # Load best_params (new) OR reconstruct from model (old ckpts)
            if 'best_params' in ckpt:
                best_params_dict = ckpt['best_params']
            else:
                model = ckpt['model']
                best_params_dict = {k: model.get_params()[k] for k in param_keys[m_lower]}
            
            params_str = ', '.join(f"{k}={v}" for k, v in sorted(best_params_dict.items()))
            
            best_hyper_table.append({
                'Feature Set': fs_display[fs],
                'Model': m_disp,
                'Best Threshold': best_t,
                'Best Hyperparameters': params_str
            })

hyper_df = pd.DataFrame(best_hyper_table)
hyper_df = hyper_df.set_index(['Feature Set', 'Model'])
print("\n🔥 **Best Hyperparameters Summary** 🔥")
display(hyper_df.style
    .set_caption("**Best Threshold & Hyperparameters per Model/Feature Set**")
    .background_gradient(cmap='YlGnBu', subset=['Best Threshold'])
    .format({'Best Threshold': '{:.3f}'})
    .set_table_styles([{'selector': 'th', 'props': [('font-weight', 'bold')]}]) )
```
